```python
import pandas as pd

file_path = r"D:\ML_PROJECT_02\data\aa_dataset-tickets-multi-lang-5-2-50-version.csv"

df = pd.read_csv(
    file_path,
    encoding="utf-8",
    on_bad_lines="skip"
)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
# Check dataset information

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df[["subject", "body", "queue", "priority", "language"]].isnull().sum())

print("\nTicket Language Distribution:")
print(df["language"].value_counts())

print("\nTicket Category Distribution:")
print(df["queue"].value_counts())

print("\nTicket Priority Distribution:")
print(df["priority"].value_counts())
```

```python
# Step 4: Clean the dataset

# Keep only the columns needed for classification
df = df[["subject", "body", "queue", "priority", "language"]].copy()

# Combine subject and body into one text column
df["ticket_text"] = (
    df["subject"].fillna("").astype(str) + " " +
    df["body"].fillna("").astype(str)
).str.strip()

# Remove rows with empty ticket text or missing category
df = df.dropna(subset=["queue"])
df = df[df["ticket_text"] != ""]

# Remove duplicate tickets
df = df.drop_duplicates(subset=["ticket_text", "queue"])

# Reset the row index
df = df.reset_index(drop=True)

print("Data cleaning completed!")
print("Rows after cleaning:", len(df))
print("Missing ticket text:", df["ticket_text"].isna().sum())
print("Missing categories:", df["queue"].isna().sum())

print("\nCategory distribution:")
print(df["queue"].value_counts())

df[["ticket_text", "queue", "priority", "language"]].head()
```

```python
# Step 5: Text preprocessing and TF-IDF

from sklearn.feature_extraction.text import TfidfVectorizer

# Remove rows with missing categories or empty ticket text
df = df.dropna(subset=["queue"])
df = df[df["ticket_text"].str.strip() != ""]

# Keep only categories that have at least 2 examples
category_counts = df["queue"].value_counts()
valid_categories = category_counts[category_counts >= 2].index
df = df[df["queue"].isin(valid_categories)].copy()

# Input features and target labels
X = df["ticket_text"]
y = df["queue"]

print("Total tickets:", len(df))
print("Number of categories:", y.nunique())
print("\nTickets per category:")
print(y.value_counts())

# Convert text into numerical features
tfidf = TfidfVectorizer(
    max_features=10000,
    stop_words="english",
    ngram_range=(1, 2),
    min_df=2
)

X_tfidf = tfidf.fit_transform(X)

print("\nTF-IDF preparation completed!")
print("TF-IDF matrix shape:", X_tfidf.shape)
```

```python
# Step 6: Split the dataset

from sklearn.model_selection import train_test_split

# Split ticket text and category labels
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Dataset split completed!")
print("-" * 40)

print("Training tickets:", len(X_train))
print("Testing tickets:", len(X_test))

print("\nTraining percentage:", round(len(X_train) / len(X) * 100, 2), "%")
print("Testing percentage:", round(len(X_test) / len(X) * 100, 2), "%")
```

```python
# Step 7: Train the Machine Learning model

from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

# Create a complete text classification pipeline
model = Pipeline([
    ("tfidf", TfidfVectorizer(
        max_features=10000,
        stop_words="english",
        ngram_range=(1, 2),
        min_df=2
    )),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ))
])

# Train the model using training data
model.fit(X_train, y_train)

print("Model training completed successfully!")
print("Model: TF-IDF + Logistic Regression")
print("Training tickets:", len(X_train))
print("Testing tickets:", len(X_test))
```

```python
# Step 8: Model Evaluation

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

# Predict ticket categories for the test dataset
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("MODEL EVALUATION")
print("=" * 50)
print(f"Accuracy: {accuracy * 100:.2f}%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
# Confusion Matrix

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=model.classes_
)

display_matrix = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=model.classes_
)

fig, ax = plt.subplots(figsize=(12, 10))
display_matrix.plot(
    ax=ax,
    xticks_rotation=90,
    colorbar=False
)

plt.title("Support Ticket Classification - Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
# Step 9: Predict the category of a new support ticket

def predict_ticket_category(subject, message):
    ticket_text = subject + " " + message

    prediction = model.predict([ticket_text])[0]

    print("SUPPORT TICKET PREDICTION")
    print("=" * 40)
    print("Subject:", subject)
    print("Message:", message)
    print("Predicted Category:", prediction)

    return prediction


# Example 1
predict_ticket_category(
    "Login problem",
    "I cannot log in to my account even after resetting my password."
)

print()

# Example 2
predict_ticket_category(
    "Payment issue",
    "I was charged twice for the same order."
)

print()

# Example 3
predict_ticket_category(
    "Technical problem",
    "The application keeps crashing whenever I try to open it."
)
```

```python
# Step 10: Prepare data for priority classification

import pandas as pd

# Check available priority labels
print("Available priority levels:")
print(df["priority"].value_counts(dropna=False))

# Keep records with valid ticket text and priority
priority_df = df.dropna(
    subset=["ticket_text", "priority"]
).copy()

priority_df["ticket_text"] = (
    priority_df["ticket_text"].astype(str).str.strip()
)

priority_df = priority_df[
    priority_df["ticket_text"] != ""
]

# Remove duplicate ticket texts
priority_df = priority_df.drop_duplicates(
    subset=["ticket_text"]
)

print("\nDataset prepared for priority classification.")
print("Total tickets:", len(priority_df))

print("\nPriority distribution:")
print(priority_df["priority"].value_counts())
```

```python
# Step 11: Train the Priority Classification Model

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Define input and target
X_priority = priority_df["ticket_text"]
y_priority = priority_df["priority"]

# Check whether there are at least two priority classes
if y_priority.nunique() < 2:
    print("Cannot train: the dataset has fewer than two priority classes.")
else:
    # Split the dataset
    Xp_train, Xp_test, yp_train, yp_test = train_test_split(
        X_priority,
        y_priority,
        test_size=0.20,
        random_state=42,
        stratify=y_priority
    )

    # Build the model
    priority_model = Pipeline([
        ("tfidf", TfidfVectorizer(
            max_features=10000,
            ngram_range=(1, 2),
            min_df=2,
            stop_words="english"
        )),
        ("classifier", LogisticRegression(
            max_iter=1000,
            class_weight="balanced"
        ))
    ])

    # Train
    priority_model.fit(Xp_train, yp_train)

    # Evaluate
    yp_pred = priority_model.predict(Xp_test)

    print("Priority model trained successfully!")
    print(f"Accuracy: {accuracy_score(yp_test, yp_pred) * 100:.2f}%")

    print("\nClassification Report:")
    print(classification_report(
        yp_test,
        yp_pred,
        zero_division=0
    ))
```

```python
# Step 12: Combined Support Ticket Prediction

def classify_support_ticket(subject, message):
    ticket_text = subject + " " + message

    # Predict support department
    predicted_category = model.predict([ticket_text])[0]

    # Predict priority
    predicted_priority = priority_model.predict([ticket_text])[0]

    print("=" * 50)
    print("SUPPORT TICKET CLASSIFICATION")
    print("=" * 50)

    print("Subject:", subject)
    print("Message:", message)
    print("-" * 50)
    print("Predicted Department:", predicted_category)
    print("Predicted Priority:", predicted_priority)
    print("=" * 50)

    return {
        "subject": subject,
        "message": message,
        "predicted_department": predicted_category,
        "predicted_priority": predicted_priority
    }


# Example prediction
result = classify_support_ticket(
    "Payment problem",
    "I was charged twice for my subscription and need a refund."
)
```
